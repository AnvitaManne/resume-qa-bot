````python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
print("Libraries imported successfully.")
````

````python
# Load environment variables from .env file
if load_dotenv():
    print("Environment variables loaded from .env file.")
else:
    print("No .env file found or it's empty. Make sure GROQ_API_KEY is set in a .env file in the same directory as this notebook.")

groq_api_key = os.getenv("GROQ_API_KEY")

chat_model = None

if not groq_api_key:
    print("ERROR: GROQ_API_KEY not found in environment variables.")
    print("Please ensure it's set in your .env file (e.g., GROQ_API_KEY='your_key_here').")
else:
    print("GROQ_API_KEY loaded successfully.")
    # Initialize the Groq Chat client
    try:
        chat_model = ChatGroq(
            groq_api_key=groq_api_key,
            model_name="llama3-8b-8192", 
            temperature=0.7 # Adjust for creativity (higher) vs. factuality (lower)
        )
        print(f"Groq Chat client initialized with model: {chat_model.model_name}")
    except Exception as e:
        print(f"Error initializing Groq client: {e}")
        print("Please check your API key and model name.")
````

````python
qa_chain = None

if chat_model: # Proceed only if chat_model was initialized successfully
    # Define a prompt template
    prompt_template = ChatPromptTemplate.from_messages([
        ("system", "You are a helpful assistant. Answer the user's questions clearly and concisely."),
        ("human", "{question}") 
    ])

    # Define the output parser
    output_parser = StrOutputParser()

    # Create the chain: Prompt -> LLM -> Output Parser
    qa_chain = prompt_template | chat_model | output_parser

    print("LangChain QA chain created successfully.")
else:
    print("Skipping LangChain pipeline creation as Groq client (chat_model) was not initialized.")
````

````python
if qa_chain:
    sample_question = "What is the colour of the sky?"
    print(f"\nTesting with sample question: '{sample_question}'")
    print("Bot: Thinking...")
    try:
        response = qa_chain.invoke({"question": sample_question})
        print("\nLLM Response:")
        print(response)
    except Exception as e:
        print(f"Error invoking QA chain: {e}")
else:
    print("Cannot test QA chain as it was not created (chat_model or chain setup failed).")
````

````python

!pip install pdfplumber faiss-cpu sentence-transformers openai

import pdfplumber
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
import openai

# Set your OpenAI key if you're using OpenAI for LLM generation
openai.api_key = "sk-..."  # replace with your API key


def extract_text_from_pdf(pdf_path):
    full_text = ""
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            full_text += page.extract_text() + "\n"
    return full_text


def chunk_text(text, chunk_size=500):
    lines = text.split("\n")
    chunks = []
    chunk = ""
    for line in lines:
        if len(chunk) + len(line) < chunk_size:
            chunk += line + "\n"
        else:
            chunks.append(chunk.strip())
            chunk = line + "\n"
    if chunk:
        chunks.append(chunk.strip())
    return chunks

def build_faiss_index(chunks, model):
    embeddings = model.encode(chunks)
    dimension = embeddings.shape[1]
    index = faiss.IndexFlatL2(dimension)
    index.add(np.array(embeddings))
    return index, embeddings, chunks

def get_top_k_chunks(question, model, index, chunks, embeddings, k=3):
    question_embedding = model.encode([question])
    D, I = index.search(np.array(question_embedding), k)
    return [chunks[i] for i in I[0]]


def generate_answer(question, context):
    prompt = f"Answer the question based ONLY on this context:\n\n{context}\n\nQuestion: {question}"
    response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You're a helpful assistant for resumes."},
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content.strip()

memory = {}


def resume_qa_bot(pdf_path, question):
    if "model" not in memory:
        memory["text"] = extract_text_from_pdf(pdf_path)
        memory["chunks"] = chunk_text(memory["text"])
        memory["model"] = SentenceTransformer('all-MiniLM-L6-v2')
        memory["index"], memory["embeddings"], memory["chunk_data"] = build_faiss_index(memory["chunks"], memory["model"])

    top_chunks = get_top_k_chunks(question, memory["model"], memory["index"], memory["chunk_data"], memory["embeddings"])
    context = "\n\n".join(top_chunks)
    return generate_answer(question, context)
````

````python
import ipywidgets as widgets
from IPython.display import display, clear_output
import io
import pdfplumber
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer
from transformers import pipeline
from collections import deque
import json


resume_text_global = ""
faiss_index_global = None
text_chunks_global = []
embedder_global = None
qa_pipeline_global = None


uploader = widgets.FileUpload(
    accept='.pdf',
    multiple=False,
    description='Upload Resume PDF'
)

process_button = widgets.Button(
    description="Process Uploaded Resume",
    button_style='info',
    tooltip='Click to extract text, create embeddings, and build FAISS index'
)
output_area = widgets.Output()


display(uploader)
display(process_button)
display(output_area)


def extract_text_from_upload(uploader_widget):
    with output_area:
        if not uploader_widget.value:
            print("No file uploaded yet. Please upload a PDF.")
            return ""
        try:
            if not isinstance(uploader_widget.value, tuple) or not uploader_widget.value:
                print("File upload data is not in the expected format. Please try uploading again.")
                return ""
            uploaded_file_dict = uploader_widget.value[0]
            if not isinstance(uploaded_file_dict, dict):
                print("Uploaded file item is not a dictionary. Please try uploading again.")
                return ""
            uploaded_filename = uploaded_file_dict.get('name', 'unknown_file.pdf')
            content = uploaded_file_dict.get('content')
            if content is None:
                print("Could not access file content. Please try uploading again.")
                return ""
        except IndexError:
            print("File upload error: No file data found. Please try uploading again.")
            return ""
        except KeyError as e:
            print(f"Could not access expected key '{e}' in file data. Please try uploading again.")
            return ""
        except Exception as e:
            print(f"An unexpected error occurred while accessing uploaded file data: {e}")
            return ""

        print(f"Extracting text from '{uploaded_filename}'...")
        all_text = ""
        try:
            with pdfplumber.open(io.BytesIO(content)) as pdf:
                for i, page in enumerate(pdf.pages):
                    page_text = page.extract_text()
                    if page_text:
                        all_text += page_text + "\n"
            if not all_text.strip():
                print("Could not extract text from the PDF.")
                return ""
            print("Text extracted successfully from PDF.")
            return all_text
        except Exception as e:
            print(f"Error processing PDF: {e}")
            return ""

def split_text(text, chunk_size=120, overlap=40):
    words = text.split()
    if not words:
        return []
    chunks = []
    for i in range(0, len(words), chunk_size - overlap):
        chunk = " ".join(words[i:i + chunk_size])
        chunks.append(chunk)
    chunks = [c for c in chunks if len(c.split()) > 10]
    if chunks:
        print(f"Split text into {len(chunks)} chunks.")
    else:
        print("No valid chunks created from the text.")
    return chunks

def create_faiss_index(text_chunks_local, embedder_local):
    if not text_chunks_local:
        print("No text chunks to index.")
        return None, None
    print(f"Creating embeddings for {len(text_chunks_local)} chunks...")
    try:
        vectors = embedder_local.encode(text_chunks_local, show_progress_bar=False)
        if not hasattr(vectors, 'shape') or vectors.shape[0] == 0:
            print("No vectors were generated from text chunks.")
            return None, None
        index = faiss.IndexFlatL2(vectors.shape[1])
        index.add(np.array(vectors, dtype=np.float32))
        print("FAISS index created successfully.")
        return index, vectors
    except Exception as e:
        print(f"Error creating FAISS index: {e}")
        return None, None

def get_answer_from_resume_qa_with_memory_handling(retrieval_query, qa_question, faiss_idx, all_chunks, embedder_local, qa_model, top_k_retrieval=3, conversation_history_for_context=None):
    if faiss_idx is None or not all_chunks:
        return "The resume has not been processed yet or contains no text."
    if not retrieval_query.strip() or not qa_question.strip():
        return "Please ask a question."

    q_vector = embedder_local.encode([retrieval_query])
    try:
        distances, indices = faiss_idx.search(np.array(q_vector, dtype=np.float32), k=min(top_k_retrieval, len(all_chunks)))
    except Exception as e:
        return f"Error during FAISS search: {e}"

    retrieved_chunk_texts = [all_chunks[i] for i in indices[0] if i < len(all_chunks)]
    if not retrieved_chunk_texts:
        return "Could not find relevant context in the resume for your question."

    base_context_from_retrieval = " ".join(retrieved_chunk_texts)
    final_context_for_qa = base_context_from_retrieval

    if conversation_history_for_context:
        history_str = "Conversation History:\n"
        for q, a in conversation_history_for_context:
            history_str += f"User: {q}\nBot: {a}\n"
        history_str += "Current Document Context:\n"
        final_context_for_qa = history_str + base_context_from_retrieval

    if not final_context_for_qa.strip():
        return "Retrieved context is empty."

    try:
        result = qa_model(question=qa_question, context=final_context_for_qa)
        if result['score'] < 0.05:
             return f"I'm not very confident (score: {result['score']:.2f}), but: {result['answer']}"
        return result['answer']
    except Exception as e:
        if "too long" in str(e).lower() or (hasattr(e, 'args') and any("too long" in str(arg).lower() for arg in e.args)):
            print(f"Context (length {len(final_context_for_qa)}) might be too long. Trying with base retrieved context only.")
            try:
                result_no_hist = qa_model(question=qa_question, context=base_context_from_retrieval)
                if result_no_hist['score'] < 0.05:
                    return f"I'm not very confident (score: {result_no_hist['score']:.2f}), but (without history): {result_no_hist['answer']}"
                return result_no_hist['answer'] + " (answered without history due to length)"
            except Exception as e2:
                print(f"Base context also too long. Truncating. Error: {e2}")
                max_len = qa_model.tokenizer.model_max_length if hasattr(qa_model, 'tokenizer') else 512
                estimated_max_chars = max_len * 3
                context_truncated = base_context_from_retrieval[:estimated_max_chars]
                try:
                    result_trunc = qa_model(question=qa_question, context=context_truncated)
                    if result_trunc['score'] < 0.05:
                        return f"I'm not very confident (score: {result_trunc['score']:.2f}), but (truncated): {result_trunc['answer']}"
                    return result_trunc['answer'] + " (from truncated context)"
                except Exception as e3:
                    return f"Error with QA model after truncating: {e3}"
        print(f"Error during QA model inference: {e}")
        return "Sorry, I encountered an error trying to answer that precisely."


def on_process_button_clicked(b):
    global resume_text_global, faiss_index_global, text_chunks_global, embedder_global, qa_pipeline_global

    with output_area:
        clear_output(wait=True)
        print("Starting resume processing...")

        if embedder_global is None:
            print("Loading sentence transformer model...")
            try:
                embedder_global = SentenceTransformer('all-MiniLM-L6-v2')
                print("Sentence transformer loaded.")
            except Exception as e:
                print(f"Failed to load Sentence Transformer: {e}")
                return
        if qa_pipeline_global is None:
            print("Loading Question Answering pipeline...")
            try:
                qa_pipeline_global = pipeline("question-answering", model="distilbert-base-cased-distilled-squad", device=-1)
                print("QA pipeline loaded.")
            except Exception as e:
                print(f"Failed to load QA pipeline: {e}")
                return

        resume_text_global = extract_text_from_upload(uploader)
        if not resume_text_global:
            print("Stopping processing: no text was extracted.")
            return

        text_chunks_global = split_text(resume_text_global)
        if not text_chunks_global:
            print("Stopping processing: no chunks were created.")
            return

        faiss_index_global, _ = create_faiss_index(text_chunks_global, embedder_global)
        if faiss_index_global is None:
            print("Stopping processing: FAISS index creation failed.")
            return

        print("\nResume processed successfully! You can now ask questions with memory in the next cell.")

process_button.on_click(on_process_button_clicked)
````

````python
from collections import deque

def start_qa_session_with_memory(max_history_len=3):
    global faiss_index_global, text_chunks_global, embedder_global, qa_pipeline_global, resume_text_global, output_area

    with output_area:
        clear_output(wait=True)
        if not resume_text_global:
            print("No resume has been processed. Please upload a PDF and click 'Process Uploaded Resume' in Cell 1.")
            return
        if faiss_index_global is None or not text_chunks_global or embedder_global is None or qa_pipeline_global is None:
            print("Resume processing was incomplete or models are not loaded. Please re-process using Cell 1.")
            return

        print("Resume processed! Ask questions below. Type 'exit' to quit.")
        print("(This version includes basic conversation memory).")

    conversation_history = deque(maxlen=max_history_len)

    while True:
        try:
            current_query = input("Ask about the resume: ")
            if current_query.lower() == "exit":
                with output_area:
                    clear_output(wait=True)
                    print("Session ended.")
                break
            if not current_query.strip():
                print("Bot: Please type a question.")
                continue

            effective_query_for_retrieval = current_query
            if conversation_history:
                if any(pronoun in current_query.lower() for pronoun in ["it", "them", "that", "those", "their", "his", "her"]):
                    last_q, last_a = conversation_history[-1]
                    effective_query_for_retrieval = f"{last_q} {last_a} {current_query}"

            answer = get_answer_from_resume_qa_with_memory_handling(
                retrieval_query=effective_query_for_retrieval,
                qa_question=current_query,
                faiss_idx=faiss_index_global,
                all_chunks=text_chunks_global,
                embedder_local=embedder_global,
                qa_model=qa_pipeline_global,
                conversation_history_for_context=list(conversation_history)
            )

            print(f"Bot: {answer}")
            if answer and not answer.startswith("Could not find") and not answer.startswith("Sorry") and not "not very confident" in answer :
                conversation_history.append((current_query, answer))

        except EOFError:
            with output_area:
                clear_output(wait=True)
                print("\nInput stream closed. Exiting Q&A session.")
            break
        except KeyboardInterrupt:
            with output_area:
                clear_output(wait=True)
                print("\nQ&A session interrupted by user. Exiting.")
            break
        except Exception as e:
            with output_area:
                clear_output(wait=True)
                print(f"An unexpected error occurred in Q&A: {e}")
            break


if resume_text_global and faiss_index_global:
    start_qa_session_with_memory()
else:
    with output_area:
        clear_output(wait=True)
        print("Please upload a PDF and click 'Process Uploaded Resume' in Cell 1, then run this cell (Cell 2) again.")
````

````python
from transformers import pipeline
import json
import ipywidgets as widgets # Already imported in Cell 1, but good for standalone cell
from IPython.display import display, clear_output # Same as above


extractor_pipeline_global = None

def initialize_extractor_model():
    global extractor_pipeline_global
    if extractor_pipeline_global is None:
        print("Initializing information extractor model (e.g., Flan-T5)...")
        try:
            extractor_pipeline_global = pipeline("text2text-generation", model="google/flan-t5-base", device=-1)
            print("Extractor model initialized.")
        except Exception as e:
            print(f"Failed to initialize extractor model: {e}")
            extractor_pipeline_global = None
    return extractor_pipeline_global

def build_extraction_prompt(resume_plain_text):
    prompt = f"""
Given the following resume text, extract the person's name, a list of their skills, and a list of their projects (with title and a short description for each).
Format the output as a JSON object with keys "name", "skills", and "projects".
Each project in the "projects" list should be an object with "title" and "description" keys.

Resume Text:
---
{resume_plain_text}
---

JSON Output:
"""
    return prompt

def extract_information_from_text(resume_plain_text):
    global extractor_pipeline_global
    if extractor_pipeline_global is None:
        print("Extractor model not initialized.")
        return None

    prompt = build_extraction_prompt(resume_plain_text)
    try:
        print("Extracting information using the language model...")
        generated_output = extractor_pipeline_global(prompt, max_length=1024, num_beams=3, early_stopping=True)
        
        if not generated_output or not generated_output[0]['generated_text']:
            print("Model did not generate any output.")
            return None

        extracted_json_str = generated_output[0]['generated_text']
        
        try:
            if extracted_json_str.strip().startswith("```json"):
                extracted_json_str = extracted_json_str.strip()[7:]
            if extracted_json_str.strip().endswith("```"):
                extracted_json_str = extracted_json_str.strip()[:-3]
            
            start_index = extracted_json_str.find('{')
            end_index = extracted_json_str.rfind('}')
            if start_index != -1 and end_index != -1 and end_index > start_index:
                json_candidate = extracted_json_str[start_index : end_index+1]
                extracted_data = json.loads(json_candidate)
                print("Successfully extracted and parsed JSON data.")
                return extracted_data
            else:
                print("Could not find a valid JSON structure in the model's output. Raw output was:")
                print(extracted_json_str)
                return {"error": "Failed to parse JSON from model output", "raw_output": extracted_json_str}

        except json.JSONDecodeError as e:
            print(f"Failed to parse JSON from model output: {e}. Raw output was:")
            print(extracted_json_str)
            return {"error": f"JSONDecodeError: {e}", "raw_output": extracted_json_str}
        except Exception as e_gen:
            print(f"An unexpected error occurred during JSON parsing or handling: {e_gen}. Raw output was:")
            print(extracted_json_str)
            return {"error": f"Unexpected parsing error: {e_gen}", "raw_output": extracted_json_str}
    except Exception as e:
        print(f"Error during information extraction: {e}")
        if "too long" in str(e).lower():
            print("The resume text might be too long for the model's context window.")
        return None


extract_info_button = widgets.Button(
    description="Extract Structured Info from Resume",
    button_style='success',
    tooltip='Click to extract Name, Skills, Projects from the processed resume text'
)
extraction_output_area = widgets.Output()

def on_extract_info_button_clicked(b):
    global resume_text_global

    with extraction_output_area:
        clear_output(wait=True)
        print("Starting structured information extraction...")

        if not resume_text_global:
            print("No resume text available. Please process a resume in Cell 1 first.")
            return

        if initialize_extractor_model() is None:
             return

        extracted_info = extract_information_from_text(resume_text_global)

        if extracted_info:
            if "error" in extracted_info:
                print("\nExtraction Attempted (with issues):")
            else:
                print("\nExtracted Information:")
            print(json.dumps(extracted_info, indent=2))
        else:
            print("Could not extract information.")

extract_info_button.on_click(on_extract_info_button_clicked)

if __name__ == '__main__' and '__file__' not in globals():
    print("\nTask 2: Resume Information Extractor")
    print("Ensure a resume has been processed in Cell 1.")
    print("Then click the button below to extract structured information.")
    display(extract_info_button)
    display(extraction_output_area)
````

````python
import ipywidgets as widgets
from IPython.display import display, clear_output
import io
import pdfplumber
import re
from typing import List, Dict, Any, Optional
from pydantic import BaseModel, Field
import os

# Import LangChain components
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.memory import ConversationBufferMemory
from langchain.chains import LLMChain, ConversationalRetrievalChain
from langchain.prompts import PromptTemplate
from langchain.llms import ChatGroq  # Using ChatGroq for better RAG performance

# Pydantic models for structured data extraction
class ContactInfo(BaseModel):
    name: Optional[str] = Field(None, description="Full name of the candidate")
    email: Optional[str] = Field(None, description="Email address of the candidate")
    phone: Optional[str] = Field(None, description="Phone number of the candidate")
    location: Optional[str] = Field(None, description="Location/address of the candidate")
    linkedin: Optional[str] = Field(None, description="LinkedIn profile URL")

class Education(BaseModel):
    institution: str = Field(..., description="Name of educational institution")
    degree: str = Field(..., description="Degree obtained or pursuing")
    field_of_study: Optional[str] = Field(None, description="Field or major of study")
    dates: Optional[str] = Field(None, description="Start and end dates of education")
    gpa: Optional[str] = Field(None, description="GPA if mentioned")

class Experience(BaseModel):
    company: str = Field(..., description="Company or organization name")
    title: str = Field(..., description="Job title or position")
    dates: Optional[str] = Field(None, description="Employment duration")
    responsibilities: List[str] = Field(default_factory=list, description="Key responsibilities and achievements")

class Skills(BaseModel):
    technical: List[str] = Field(default_factory=list, description="Technical skills (programming, tools, etc.)")
    soft: List[str] = Field(default_factory=list, description="Soft skills (communication, leadership, etc.)")
    languages: List[str] = Field(default_factory=list, description="Language proficiencies")

class ResumeData(BaseModel):
    contact_info: ContactInfo = Field(default_factory=ContactInfo)
    summary: Optional[str] = Field(None, description="Professional summary or objective statement")
    education: List[Education] = Field(default_factory=list)
    experience: List[Experience] = Field(default_factory=list)
    skills: Skills = Field(default_factory=Skills)
    certifications: List[str] = Field(default_factory=list, description="Professional certifications")

# State variables
processed_resumes = {}
resume_memories = {}
resume_structured_data = {}
vectorstores = {}

current_chat_resume_name = None
lc_memory_for_current_session = None
retrieval_chain = None
llm = None
embeddings = None

# Initialize embeddings model
def initialize_embeddings():
    global embeddings
    if embeddings is None:
        try:
            embeddings = HuggingFaceEmbeddings(
                model_name="sentence-transformers/all-MiniLM-L6-v2",
                model_kwargs={'device': 'cpu'}
            )
            return embeddings
        except Exception as e:
            print(f"Error initializing embeddings model: {e}")
            return None
    return embeddings

# Initialize LLM for RAG
def initialize_llm():
    global llm
    if llm is None:
        try:
            llm = ChatGroq(
                model_name="llama2-70b",
                temperature=0.2,
                max_tokens=512
            )
            return llm
        except Exception as e:
            print(f"Error initializing LLM: {e}")
            print("Falling back to open-source HuggingFace model...")
            try:
                from langchain.llms import HuggingFaceEndpoint
                llm = HuggingFaceEndpoint(
                    repo_id="google/flan-t5-large",  # Using larger model than base
                    temperature=0.1,
                    max_length=512
                )
                return llm
            except Exception as e2:
                print(f"Error initializing fallback LLM: {e2}")
                return None
    return llm

# Helper: Process resume file
def extract_and_process_pdf_data(file_name, file_content):
    try:
        # Extract text from PDF
        with pdfplumber.open(io.BytesIO(file_content)) as pdf:
            all_text = "\n".join([page.extract_text() or "" for page in pdf.pages])
        
        if not all_text.strip():
            return None
        
        # Split text into chunks for vectorization
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=200,
            length_function=len
        )
        chunks = text_splitter.split_text(all_text)
        
        # Get embeddings
        emb = initialize_embeddings()
        if not emb:
            return None
            
        # Create FAISS vector store
        vectorstore = FAISS.from_texts(chunks, emb)
        
        # Try to extract structured data (using LLM if available)
        structured_data = extract_structured_data(all_text)
        
        return {
            "text": all_text, 
            "chunks": chunks,
            "structured_data": structured_data
        }
    except Exception as e:
        print(f"Error processing PDF: {e}")
        return None

# Helper: Extract structured data using Pydantic schema
def extract_structured_data(text):
    # Initialize a basic ResumeData object
    resume_data = ResumeData()
    
    # Extract contact info using regex patterns
    # Name extraction - look for name at the beginning of resume
    name_match = re.search(r'^([A-Z][a-z]+(?:\s[A-Z][a-z]+){1,3})', text, re.MULTILINE)
    if name_match:
        resume_data.contact_info.name = name_match.group(1)
    
    # Email extraction
    email_match = re.search(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)
    if email_match:
        resume_data.contact_info.email = email_match.group(0)
    
    # Phone extraction
    phone_match = re.search(r'\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}', text)
    if phone_match:
        resume_data.contact_info.phone = phone_match.group(0)
    
    # LinkedIn extraction
    linkedin_match = re.search(r'linkedin\.com/in/[a-zA-Z0-9_-]+', text)
    if linkedin_match:
        resume_data.contact_info.linkedin = "https://" + linkedin_match.group(0)
    
    # Extract education based on common degree indicators
    degrees = ['Bachelor', 'Master', 'PhD', 'B.S.', 'M.S.', 'B.A.', 'M.A.', 'MBA']
    education_sections = []
    
    for degree in degrees:
        pattern = fr'{degree}[^.]*?(?:\d{{4}}|University|College|Institute)[^.]*?\.'
        matches = re.finditer(pattern, text, re.IGNORECASE)
        for match in matches:
            edu_text = match.group(0)
            education_sections.append(Education(
                institution=re.search(r'(?:University|College|Institute)[^.,]*', edu_text, re.IGNORECASE).group(0) if re.search(r'(?:University|College|Institute)[^.,]*', edu_text, re.IGNORECASE) else "Unknown Institution",
                degree=degree,
                dates=re.search(r'\d{4}(?:\s*-\s*\d{4}|\s*-\s*Present|\s*-\s*Current)?', edu_text).group(0) if re.search(r'\d{4}(?:\s*-\s*\d{4}|\s*-\s*Present|\s*-\s*Current)?', edu_text) else None
            ))
    
    resume_data.education = education_sections
    
    # Extract technical skills by looking for common programming languages and tools
    tech_skills = []
    tech_keywords = ['Python', 'Java', 'JavaScript', 'SQL', 'C++', 'HTML', 'CSS', 'React', 
                    'Angular', 'Node.js', 'AWS', 'Docker', 'Kubernetes', 'Git', 'Machine Learning',
                    'Data Science', 'TensorFlow', 'PyTorch', 'Excel', 'PowerPoint']
    
    # Check for each keyword in the text
    for skill in tech_keywords:
        if re.search(r'\b' + re.escape(skill) + r'\b', text, re.IGNORECASE):
            tech_skills.append(skill)
    
    resume_data.skills.technical = tech_skills
    
    return resume_data

# UI: Create and display widgets
uploader = widgets.FileUpload(accept='.pdf', multiple=False, description="Upload Resume")
process_button = widgets.Button(description="Process Uploaded Resume", button_style='info')
output_area = widgets.Output()
resume_selector = widgets.Dropdown(description='Chat with:')
start_chat_button = widgets.Button(description="Start/Switch Chat", button_style='success')
chat_output_area = widgets.Output(layout={'border': '1px solid black', 'height': '300px', 'overflow_y': 'scroll'})
chat_input_text = widgets.Text(placeholder='Type question or "exit"', description='You:', layout=widgets.Layout(width='auto'), disabled=True)
send_chat_button = widgets.Button(description="Send", button_style='primary', disabled=True)
show_full_chat_button = widgets.Button(description="Show Full Chat History", button_style='warning')
show_resume_info_button = widgets.Button(description="Show Resume Details", button_style='info')

# Helper: Update dropdown
def update_resume_selector_options():
    current_value = resume_selector.value
    resume_selector.options = list(processed_resumes.keys())
    if not processed_resumes:
        resume_selector.disabled = True
        start_chat_button.disabled = True
        chat_input_text.disabled = True
        send_chat_button.disabled = True
    else:
        resume_selector.disabled = False
        start_chat_button.disabled = False
        resume_selector.value = current_value if current_value in processed_resumes else resume_selector.options[0]

def reset_uploader():
    global uploader
    uploader.value = ()
    uploader._counter = 0

# UI: Display chat
def refresh_chat_display_lc():
    with chat_output_area:
        clear_output(wait=True)
        if current_chat_resume_name and lc_memory_for_current_session:
            print(f"--- Chatting with: {current_chat_resume_name} ---")
            for msg in lc_memory_for_current_session.chat_memory.messages:
                role = "You" if msg.type == "human" else "Bot"
                print(f"{role}: {msg.content}\n{'-'*20}")
        elif current_chat_resume_name == "ENDED_SESSION":
            print("Chat session ended. Upload or select a resume.")
        else:
            print("No active chat. Select a resume to begin.")

# Process resume button handler
def on_process_button_clicked(b):
    with output_area:
        clear_output()
        if not uploader.value:
            print("No file uploaded.")
            return
        
        uploaded_files = uploader.value
        if isinstance(uploaded_files, tuple):
            uploaded = uploaded_files[0]
        elif isinstance(uploaded_files, dict):
            uploaded = list(uploaded_files.values())[0]
        else:
            print("Unexpected uploader.value type:", type(uploaded_files))
            return
        
        if 'metadata' in uploaded and 'name' in uploaded['metadata']:
            file_name = uploaded['metadata']['name']
        elif 'name' in uploaded:
            file_name = uploaded['name']
        else:
            print("Can't find filename in uploaded data.")
            return
        
        if 'content' in uploaded:
            content = uploaded['content']
        else:
            print("Can't find file content in uploaded data.")
            return
        
        if file_name in processed_resumes:
            print(f"'{file_name}' already processed.")
        else:
            print(f"Processing '{file_name}'...")
            data = extract_and_process_pdf_data(file_name, content)
            if data:
                processed_resumes[file_name] = data
                
                # Create vector store for the resume
                if initialize_embeddings():
                    try:
                        vectorstore = FAISS.from_texts(data["chunks"], initialize_embeddings())
                        vectorstores[file_name] = vectorstore
                        resume_structured_data[file_name] = data.get("structured_data")
                        print(f"Processed '{file_name}' successfully. Vector store created.")
                    except Exception as e:
                        print(f"Error creating vector store: {e}")
                else:
                    print("Warning: Embedding model not available. Vector search disabled.")
            else:
                print("Failed to extract data.")
        
        update_resume_selector_options()
        reset_uploader()

process_button.on_click(on_process_button_clicked)

# Button: Start Chat
def on_start_chat_button_clicked(b):
    global current_chat_resume_name, lc_memory_for_current_session, retrieval_chain

    selected_resume = resume_selector.value
    if not selected_resume or selected_resume not in processed_resumes:
        return

    # Initialize LLM
    if initialize_llm() is None:
        with chat_output_area:
            clear_output()
            print("Failed to initialize LLM. Check your connections and API keys.")
        return

    current_chat_resume_name = selected_resume

    # Set up or retrieve memory
    if selected_resume in resume_memories:
        lc_memory_for_current_session = resume_memories[selected_resume]
    else:
        lc_memory_for_current_session = ConversationBufferMemory(
            memory_key="chat_history",
            return_messages=True
        )
        resume_memories[selected_resume] = lc_memory_for_current_session
    
    # Create retrieval chain with the vectorstore
    if selected_resume in vectorstores:
        try:
            # Set up the ConversationalRetrievalChain
            retrieval_chain = ConversationalRetrievalChain.from_llm(
                llm=initialize_llm(),
                retriever=vectorstores[selected_resume].as_retriever(
                    search_kwargs={"k": 3}  # Retrieve top 3 most relevant chunks
                ),
                memory=lc_memory_for_current_session,
                verbose=True
            )
        except Exception as e:
            print(f"Error creating retrieval chain: {e}")
            # Fallback to a simple LLM chain if retrieval chain fails
            template = """Use the following pieces of context and chat history to answer the question. 
            If you don't know the answer based on the context, just say that you don't know.
            
            Context: 
            {context}
            
            Chat History:
            {chat_history}
            
            Question: {question}
            Answer:"""
            
            prompt = PromptTemplate(
                input_variables=["context", "chat_history", "question"],
                template=template
            )
            
            retrieval_chain = LLMChain(
                llm=initialize_llm(),
                prompt=prompt,
                memory=lc_memory_for_current_session
            )
    
    refresh_chat_display_lc()
    chat_input_text.disabled = False
    send_chat_button.disabled = False
    chat_input_text.value = ""
    chat_input_text.focus()

start_chat_button.on_click(on_start_chat_button_clicked)

# Button: Send Chat
def on_send_chat_button_clicked(b):
    global current_chat_resume_name

    query = chat_input_text.value.strip()
    if not query or not current_chat_resume_name:
        return

    if query.lower() == 'exit':
        lc_memory_for_current_session.chat_memory.add_user_message(query)
        lc_memory_for_current_session.chat_memory.add_ai_message(f"Ended chat with {current_chat_resume_name}.")
        current_chat_resume_name = "ENDED_SESSION"
        chat_input_text.disabled = True
        send_chat_button.disabled = True
        chat_input_text.value = ""
        refresh_chat_display_lc()
        return

    # Add user message to memory
    lc_memory_for_current_session.chat_memory.add_user_message(query)
    
    with chat_output_area:
        clear_output(wait=True)
        print(f"You: {query}")
        print("Bot: Thinking...")

    try:
        if current_chat_resume_name in vectorstores:
            # If using ConversationalRetrievalChain
            if hasattr(retrieval_chain, 'run'):
                answer = retrieval_chain.run(query)
            else:
                # Get relevant context from vector store
                relevant_docs = vectorstores[current_chat_resume_name].similarity_search(query, k=3)
                context = "\n".join([doc.page_content for doc in relevant_docs])
                
                # For LLMChain fallback
                answer = retrieval_chain.run(
                    context=context,
                    chat_history=lc_memory_for_current_session.chat_memory.messages,
                    question=query
                )
        else:
            # Fallback to basic content if vector store is unavailable
            resume_text = processed_resumes[current_chat_resume_name]["text"]
            answer = initialize_llm().generate(
                [f"Based on this resume: {resume_text[:2000]}... Answer this question: {query}"]
            ).generations[0][0].text
    except Exception as e:
        answer = f"I encountered an error when trying to process your question. Please try again or rephrase your question. Technical details: {str(e)}"

    # Add answer to memory
    lc_memory_for_current_session.chat_memory.add_ai_message(answer)
    
    refresh_chat_display_lc()
    chat_input_text.value = ""
    chat_input_text.focus()

send_chat_button.on_click(on_send_chat_button_clicked)

# Button: Show Full Chat History
def on_show_full_chat_clicked(b):
    with chat_output_area:
        clear_output(wait=True)
        if lc_memory_for_current_session:
            print(f" Full Chat History with: {current_chat_resume_name} ")
            for msg in lc_memory_for_current_session.chat_memory.messages:
                role = "You" if msg.type == "human" else "Bot"
                print(f"{role}: {msg.content}\n{'-'*30}")
        else:
            print("No chat history found.")

show_full_chat_button.on_click(on_show_full_chat_clicked)

# Button: Show Resume Details
def on_show_resume_info_clicked(b):
    with chat_output_area:
        clear_output(wait=True)
        selected = resume_selector.value
        
        if not selected or selected not in processed_resumes:
            print("No resume selected.")
            return
            
        print(f"=== Resume Details: {selected} ===\n")
        
        # Show structured data if available
        if selected in resume_structured_data and resume_structured_data[selected]:
            data = resume_structured_data[selected]
            
            if data.contact_info and data.contact_info.name:
                print(f"Name: {data.contact_info.name}")
            if data.contact_info and data.contact_info.email:
                print(f"Email: {data.contact_info.email}")
            if data.contact_info and data.contact_info.phone:
                print(f"Phone: {data.contact_info.phone}")
            if data.contact_info and data.contact_info.linkedin:
                print(f"LinkedIn: {data.contact_info.linkedin}")
                
            print("\n--- Education ---")
            for edu in data.education:
                print(f"• {edu.degree} at {edu.institution} {edu.dates if edu.dates else ''}")
                
            print("\n--- Skills ---")
            if data.skills and data.skills.technical:
                print("Technical: " + ", ".join(data.skills.technical))
        else:
            print("No structured data available for this resume.")
            
        print("\nUse the chat to ask specific questions about this resume.")

show_resume_info_button.on_click(on_show_resume_info_clicked)

# Display UI
display(widgets.VBox([
    widgets.HTML("<h2>Enhanced Resume Chatbot</h2>"),
    widgets.HBox([uploader, process_button]),
    output_area,
    widgets.HBox([resume_selector, start_chat_button]),
    chat_output_area,
    widgets.HBox([chat_input_text, send_chat_button]),
    widgets.HBox([show_full_chat_button, show_resume_info_button])
]))

# Add helpful instructions
print("Instructions:")
print("1. Upload a resume PDF using the upload button")
print("2. Click 'Process Uploaded Resume' to extract information")
print("3. Select the resume from the dropdown and click 'Start/Switch Chat'")
print("4. Ask questions about the resume in the chat input")
print("5. Type 'exit' to end the current chat session")
````

````python
import ipywidgets as widgets
from IPython.display import display, clear_output
import io
import pdfplumber
import re
from typing import List, Dict, Any, Optional
from pydantic import BaseModel, Field
import os

try:
    import langchain_groq
except ImportError:
    print("Installing langchain-groq...")
    !pip install langchain-groq
    import langchain_groq

# Import LangChain components
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.memory import ConversationBufferMemory
from langchain.chains import LLMChain, ConversationalRetrievalChain
from langchain.prompts import PromptTemplate
from langchain_groq import ChatGroq  # Using ChatGroq as specified in requirements

# Pydantic models for structured data extraction
class ContactInfo(BaseModel):
    name: Optional[str] = Field(None, description="Full name of the candidate")
    email: Optional[str] = Field(None, description="Email address of the candidate")
    phone: Optional[str] = Field(None, description="Phone number of the candidate")
    location: Optional[str] = Field(None, description="Location/address of the candidate")
    linkedin: Optional[str] = Field(None, description="LinkedIn profile URL")

class Education(BaseModel):
    institution: str = Field(..., description="Name of educational institution")
    degree: str = Field(..., description="Degree obtained or pursuing")
    field_of_study: Optional[str] = Field(None, description="Field or major of study")
    dates: Optional[str] = Field(None, description="Start and end dates of education")
    gpa: Optional[str] = Field(None, description="GPA if mentioned")

class Experience(BaseModel):
    company: str = Field(..., description="Company or organization name")
    title: str = Field(..., description="Job title or position")
    dates: Optional[str] = Field(None, description="Employment duration")
    responsibilities: List[str] = Field(default_factory=list, description="Key responsibilities and achievements")

class Skills(BaseModel):
    technical: List[str] = Field(default_factory=list, description="Technical skills (programming, tools, etc.)")
    soft: List[str] = Field(default_factory=list, description="Soft skills (communication, leadership, etc.)")
    languages: List[str] = Field(default_factory=list, description="Language proficiencies")

class ResumeData(BaseModel):
    contact_info: ContactInfo = Field(default_factory=ContactInfo)
    summary: Optional[str] = Field(None, description="Professional summary or objective statement")
    education: List[Education] = Field(default_factory=list)
    experience: List[Experience] = Field(default_factory=list)
    skills: Skills = Field(default_factory=Skills)
    certifications: List[str] = Field(default_factory=list, description="Professional certifications")

# State variables
processed_resumes = {}
resume_memories = {}
resume_structured_data = {}
vectorstores = {}

current_chat_resume_name = None
lc_memory_for_current_session = None
retrieval_chain = None
llm = None
embeddings = None

# Initialize embeddings model
def initialize_embeddings():
    global embeddings
    if embeddings is None:
        try:
            embeddings = HuggingFaceEmbeddings(
                model_name="sentence-transformers/all-MiniLM-L6-v2",
                model_kwargs={'device': 'cpu'}
            )
            return embeddings
        except Exception as e:
            print(f"Error initializing embeddings model: {e}")
            return None
    return embeddings

# Initialize LLM for RAG
def initialize_llm():
    global llm
    if llm is None:
        try:
            llm = ChatGroq(
                model_name="llama2-70b-chat",  
                temperature=0.2,
                max_tokens=512
            )
            return llm
        except Exception as e:
            print(f"Error initializing LLM: {e}")
            print("Falling back to open-source HuggingFace model...")
            try:
                from langchain.llms import HuggingFaceEndpoint
                print("Using HuggingFaceEndpoint as fallback...")
                llm = HuggingFaceEndpoint(
                    repo_id="google/flan-t5-large", 
                    temperature=0.1,
                    max_length=512
                )
                return llm
            except Exception as e2:
                print(f"Error initializing fallback LLM: {e2}")
                return None
    return llm

# Helper: Process resume file
def extract_and_process_pdf_data(file_name, file_content):
    try:
        # Extract text from PDF
        with pdfplumber.open(io.BytesIO(file_content)) as pdf:
            all_text = "\n".join([page.extract_text() or "" for page in pdf.pages])
        
        if not all_text.strip():
            return None
        
        # Split text into chunks for vectorization
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=200,
            length_function=len
        )
        chunks = text_splitter.split_text(all_text)
        
        # Get embeddings
        emb = initialize_embeddings()
        if not emb:
            return None
            
        # Create FAISS vector store
        vectorstore = FAISS.from_texts(chunks, emb)
        
        # Try to extract structured data (using LLM if available)
        structured_data = extract_structured_data(all_text)
        
        return {
            "text": all_text, 
            "chunks": chunks,
            "structured_data": structured_data
        }
    except Exception as e:
        print(f"Error processing PDF: {e}")
        return None

# Helper: Extract structured data using Pydantic schema
def extract_structured_data(text):
    # Initialize a basic ResumeData object
    resume_data = ResumeData()
    
    name_match = re.search(r'^([A-Z][a-z]+(?:\s[A-Z][a-z]+){1,3})', text, re.MULTILINE)
    if name_match:
        resume_data.contact_info.name = name_match.group(1)
    
    # Email extraction
    email_match = re.search(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)
    if email_match:
        resume_data.contact_info.email = email_match.group(0)
    
    # Phone extraction
    phone_match = re.search(r'\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}', text)
    if phone_match:
        resume_data.contact_info.phone = phone_match.group(0)
    
    # LinkedIn extraction
    linkedin_match = re.search(r'linkedin\.com/in/[a-zA-Z0-9_-]+', text)
    if linkedin_match:
        resume_data.contact_info.linkedin = "https://" + linkedin_match.group(0)
    
    # Extract education based on common degree indicators
    degrees = ['Bachelor', 'Master', 'PhD', 'B.S.', 'M.S.', 'B.A.', 'M.A.', 'MBA']
    education_sections = []
    
    for degree in degrees:
        pattern = fr'{degree}[^.]*?(?:\d{{4}}|University|College|Institute)[^.]*?\.'
        matches = re.finditer(pattern, text, re.IGNORECASE)
        for match in matches:
            edu_text = match.group(0)
            education_sections.append(Education(
                institution=re.search(r'(?:University|College|Institute)[^.,]*', edu_text, re.IGNORECASE).group(0) if re.search(r'(?:University|College|Institute)[^.,]*', edu_text, re.IGNORECASE) else "Unknown Institution",
                degree=degree,
                dates=re.search(r'\d{4}(?:\s*-\s*\d{4}|\s*-\s*Present|\s*-\s*Current)?', edu_text).group(0) if re.search(r'\d{4}(?:\s*-\s*\d{4}|\s*-\s*Present|\s*-\s*Current)?', edu_text) else None
            ))
    
    resume_data.education = education_sections
    
    tech_skills = []
    tech_keywords = ['Python', 'Java', 'JavaScript', 'SQL', 'C++', 'HTML', 'CSS', 'React', 
                    'Angular', 'Node.js', 'AWS', 'Docker', 'Kubernetes', 'Git', 'Machine Learning',
                    'Data Science', 'TensorFlow', 'PyTorch', 'Excel', 'PowerPoint']
    
    # Check for each keyword in the text
    for skill in tech_keywords:
        if re.search(r'\b' + re.escape(skill) + r'\b', text, re.IGNORECASE):
            tech_skills.append(skill)
    
    resume_data.skills.technical = tech_skills
    
    return resume_data

# UI
uploader = widgets.FileUpload(accept='.pdf', multiple=False, description="Upload Resume")
process_button = widgets.Button(description="Process Uploaded Resume", button_style='info')
output_area = widgets.Output()
resume_selector = widgets.Dropdown(description='Chat with:')
start_chat_button = widgets.Button(description="Start/Switch Chat", button_style='success')
chat_output_area = widgets.Output(layout={'border': '1px solid black', 'height': '300px', 'overflow_y': 'scroll'})
chat_input_text = widgets.Text(placeholder='Type question or "exit"', description='You:', layout=widgets.Layout(width='auto'), disabled=True)
send_chat_button = widgets.Button(description="Send", button_style='primary', disabled=True)
show_full_chat_button = widgets.Button(description="Show Full Chat History", button_style='warning')
show_resume_info_button = widgets.Button(description="Show Resume Details", button_style='info')

# Helper: Update dropdown
def update_resume_selector_options():
    current_value = resume_selector.value
    resume_selector.options = list(processed_resumes.keys())
    if not processed_resumes:
        resume_selector.disabled = True
        start_chat_button.disabled = True
        chat_input_text.disabled = True
        send_chat_button.disabled = True
    else:
        resume_selector.disabled = False
        start_chat_button.disabled = False
        resume_selector.value = current_value if current_value in processed_resumes else resume_selector.options[0]

def reset_uploader():
    global uploader
    uploader.value = ()
    uploader._counter = 0

# UI: Display chat
def refresh_chat_display_lc():
    with chat_output_area:
        clear_output(wait=True)
        if current_chat_resume_name and lc_memory_for_current_session:
            print(f"--- Chatting with: {current_chat_resume_name} ---")
            for msg in lc_memory_for_current_session.chat_memory.messages:
                role = "You" if msg.type == "human" else "Bot"
                print(f"{role}: {msg.content}\n{'-'*20}")
        elif current_chat_resume_name == "ENDED_SESSION":
            print("Chat session ended. Upload or select a resume.")
        else:
            print("No active chat. Select a resume to begin.")

# Process resume button handler
def on_process_button_clicked(b):
    with output_area:
        clear_output()
        if not uploader.value:
            print("No file uploaded.")
            return
        
        uploaded_files = uploader.value
        if isinstance(uploaded_files, tuple):
            uploaded = uploaded_files[0]
        elif isinstance(uploaded_files, dict):
            uploaded = list(uploaded_files.values())[0]
        else:
            print("Unexpected uploader.value type:", type(uploaded_files))
            return
        
        if 'metadata' in uploaded and 'name' in uploaded['metadata']:
            file_name = uploaded['metadata']['name']
        elif 'name' in uploaded:
            file_name = uploaded['name']
        else:
            print("Can't find filename in uploaded data.")
            return
        
        if 'content' in uploaded:
            content = uploaded['content']
        else:
            print("Can't find file content in uploaded data.")
            return
        
        if file_name in processed_resumes:
            print(f"'{file_name}' already processed.")
        else:
            print(f"Processing '{file_name}'...")
            data = extract_and_process_pdf_data(file_name, content)
            if data:
                processed_resumes[file_name] = data
                
                # Create vector store for the resume
                if initialize_embeddings():
                    try:
                        vectorstore = FAISS.from_texts(data["chunks"], initialize_embeddings())
                        vectorstores[file_name] = vectorstore
                        resume_structured_data[file_name] = data.get("structured_data")
                        print(f"Processed '{file_name}' successfully. Vector store created.")
                    except Exception as e:
                        print(f"Error creating vector store: {e}")
                else:
                    print("Warning: Embedding model not available. Vector search disabled.")
            else:
                print("Failed to extract data.")
        
        update_resume_selector_options()
        reset_uploader()

process_button.on_click(on_process_button_clicked)

# Button: Start Chat
def on_start_chat_button_clicked(b):
    global current_chat_resume_name, lc_memory_for_current_session, retrieval_chain

    selected_resume = resume_selector.value
    if not selected_resume or selected_resume not in processed_resumes:
        return

    # Initialize LLM
    if initialize_llm() is None:
        with chat_output_area:
            clear_output()
            print("Failed to initialize LLM. Check your connections and API keys.")
        return

    current_chat_resume_name = selected_resume

    # Set up or retrieve memory
    if selected_resume in resume_memories:
        lc_memory_for_current_session = resume_memories[selected_resume]
    else:
        lc_memory_for_current_session = ConversationBufferMemory(
            memory_key="chat_history",
            return_messages=True
        )
        resume_memories[selected_resume] = lc_memory_for_current_session
    
    # Create retrieval chain with the vectorstore
    if selected_resume in vectorstores:
        try:
            # Set up the ConversationalRetrievalChain
            retrieval_chain = ConversationalRetrievalChain.from_llm(
                llm=initialize_llm(),
                retriever=vectorstores[selected_resume].as_retriever(
                    search_kwargs={"k": 3}  # Retrieve top 3 most relevant chunks
                ),
                memory=lc_memory_for_current_session,
                verbose=True
            )
        except Exception as e:
            print(f"Error creating retrieval chain: {e}")
            # Fallback to a simple LLM chain if retrieval chain fails
            template = """Use the following pieces of context and chat history to answer the question. 
            If you don't know the answer based on the context, just say that you don't know.
            
            Context: 
            {context}
            
            Chat History:
            {chat_history}
            
            Question: {question}
            Answer:"""
            
            prompt = PromptTemplate(
                input_variables=["context", "chat_history", "question"],
                template=template
            )
            
            retrieval_chain = LLMChain(
                llm=initialize_llm(),
                prompt=prompt,
                memory=lc_memory_for_current_session
            )
    
    refresh_chat_display_lc()
    chat_input_text.disabled = False
    send_chat_button.disabled = False
    chat_input_text.value = ""
    chat_input_text.focus()

start_chat_button.on_click(on_start_chat_button_clicked)

# Button: Send Chat
def on_send_chat_button_clicked(b):
    global current_chat_resume_name

    query = chat_input_text.value.strip()
    if not query or not current_chat_resume_name:
        return

    if query.lower() == 'exit':
        lc_memory_for_current_session.chat_memory.add_user_message(query)
        lc_memory_for_current_session.chat_memory.add_ai_message(f"Ended chat with {current_chat_resume_name}.")
        current_chat_resume_name = "ENDED_SESSION"
        chat_input_text.disabled = True
        send_chat_button.disabled = True
        chat_input_text.value = ""
        refresh_chat_display_lc()
        return

    # Add user message to memory
    lc_memory_for_current_session.chat_memory.add_user_message(query)
    
    with chat_output_area:
        clear_output(wait=True)
        print(f"You: {query}")
        print("Bot: Thinking...")

    try:
        if current_chat_resume_name in vectorstores:
            # If using ConversationalRetrievalChain
            if hasattr(retrieval_chain, 'run'):
                answer = retrieval_chain.run(query)
            else:
                # Get relevant context from vector store
                relevant_docs = vectorstores[current_chat_resume_name].similarity_search(query, k=3)
                context = "\n".join([doc.page_content for doc in relevant_docs])
                
                # For LLMChain fallback
                answer = retrieval_chain.run(
                    context=context,
                    chat_history=lc_memory_for_current_session.chat_memory.messages,
                    question=query
                )
        else:
            # Fallback to basic content if vector store is unavailable
            resume_text = processed_resumes[current_chat_resume_name]["text"]
            answer = initialize_llm().generate(
                [f"Based on this resume: {resume_text[:2000]}... Answer this question: {query}"]
            ).generations[0][0].text
    except Exception as e:
        answer = f"I encountered an error when trying to process your question. Please try again or rephrase your question. Technical details: {str(e)}"

    # Add answer to memory
    lc_memory_for_current_session.chat_memory.add_ai_message(answer)
    
    refresh_chat_display_lc()
    chat_input_text.value = ""
    chat_input_text.focus()

send_chat_button.on_click(on_send_chat_button_clicked)

# Button: Show Full Chat History
def on_show_full_chat_clicked(b):
    with chat_output_area:
        clear_output(wait=True)
        if lc_memory_for_current_session:
            print(f" Full Chat History with: {current_chat_resume_name} ")
            for msg in lc_memory_for_current_session.chat_memory.messages:
                role = "You" if msg.type == "human" else "Bot"
                print(f"{role}: {msg.content}\n{'-'*30}")
        else:
            print("No chat history found.")

show_full_chat_button.on_click(on_show_full_chat_clicked)

# Button: Show Resume Details
def on_show_resume_info_clicked(b):
    with chat_output_area:
        clear_output(wait=True)
        selected = resume_selector.value
        
        if not selected or selected not in processed_resumes:
            print("No resume selected.")
            return
            
        print(f"=== Resume Details: {selected} ===\n")
        
        # Show structured data if available
        if selected in resume_structured_data and resume_structured_data[selected]:
            data = resume_structured_data[selected]


            
            
            if data.contact_info and data.contact_info.name:
                print(f"Name: {data.contact_info.name}")
            if data.contact_info and data.contact_info.email:
                print(f"Email: {data.contact_info.email}")
            if data.contact_info and data.contact_info.phone:
                print(f"Phone: {data.contact_info.phone}")
            if data.contact_info and data.contact_info.linkedin:
                print(f"LinkedIn: {data.contact_info.linkedin}")
                
            print("\n--- Education ---")
            for edu in data.education:
                print(f"• {edu.degree} at {edu.institution} {edu.dates if edu.dates else ''}")
                
            print("\n--- Skills ---")
            if data.skills and data.skills.technical:
                print("Technical: " + ", ".join(data.skills.technical))
        else:
            print("No structured data available for this resume.")
            
        print("\nUse the chat to ask specific questions about this resume.")

show_resume_info_button.on_click(on_show_resume_info_clicked)

# Display UI
display(widgets.VBox([
    widgets.HTML("<h2>Enhanced Resume Chatbot</h2>"),
    widgets.HBox([uploader, process_button]),
    output_area,
    widgets.HBox([resume_selector, start_chat_button]),
    chat_output_area,
    widgets.HBox([chat_input_text, send_chat_button]),
    widgets.HBox([show_full_chat_button, show_resume_info_button])
]))

# Add helpful instructions
print("Instructions:")
print("1. Upload a resume PDF using the upload button")
print("2. Click 'Process Uploaded Resume' to extract information")
print("3. Select the resume from the dropdown and click 'Start/Switch Chat'")
print("4. Ask questions about the resume in the chat input")
print("5. Type 'exit' to end the current chat session")
````

````python
import io
import pdfplumber
import re
from typing import List, Dict, Any, Optional
from pydantic import BaseModel, Field
import os
try:
    import langchain_groq
    from langchain.embeddings import HuggingFaceEmbeddings
    from langchain.vectorstores import FAISS
except ImportError:
    print("Installing required packages...")
    !pip install langchain-groq faiss-cpu sentence-transformers
    import langchain_groq
    from langchain.embeddings import HuggingFaceEmbeddings
    from langchain.vectorstores import FAISS

# Import LangChain components
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationalRetrievalChain
from langchain.prompts import PromptTemplate
from langchain_groq import ChatGroq
from langchain.schema import Document

# Set up API keys (replace with your actual API keys)
os.environ["GROQ_API_KEY"] = "your_groq_api_key_here"  

# Pydantic models for structured data extraction
class ContactInfo(BaseModel):
    """Contact information from a resume"""
    name: Optional[str] = Field(None, description="Full name of the candidate")
    email: Optional[str] = Field(None, description="Email address of the candidate")
    phone: Optional[str] = Field(None, description="Phone number of the candidate")
    location: Optional[str] = Field(None, description="Location/address of the candidate")
    linkedin: Optional[str] = Field(None, description="LinkedIn profile URL")
    
class Education(BaseModel):
    """Education information from a resume"""
    institution: str = Field(..., description="Name of educational institution")
    degree: Optional[str] = Field(None, description="Degree obtained or pursuing")
    field_of_study: Optional[str] = Field(None, description="Field or major of study")
    dates: Optional[str] = Field(None, description="Start and end dates of education")
    gpa: Optional[str] = Field(None, description="GPA if mentioned")

class Experience(BaseModel):
    """Work experience information from a resume"""
    company: str = Field(..., description="Company or organization name")
    title: str = Field(..., description="Job title or position")
    dates: Optional[str] = Field(None, description="Employment duration")
    responsibilities: List[str] = Field(default_factory=list, description="Key responsibilities and achievements")

class Skills(BaseModel):
    """Skills information from a resume"""
    technical: List[str] = Field(default_factory=list, description="Technical skills (programming, tools, etc.)")
    soft: List[str] = Field(default_factory=list, description="Soft skills (communication, leadership, etc.)")
    languages: List[str] = Field(default_factory=list, description="Language proficiencies")

class ResumeData(BaseModel):
    """Comprehensive resume data structure"""
    contact_info: ContactInfo = Field(default_factory=ContactInfo)
    summary: Optional[str] = Field(None, description="Professional summary or objective statement")
    education: List[Education] = Field(default_factory=list)
    experience: List[Experience] = Field(default_factory=list)
    skills: Skills = Field(default_factory=Skills)
    certifications: List[str] = Field(default_factory=list, description="Professional certifications")

class ResumeProcessor:
    """Core class for processing resumes with advanced retrieval capabilities"""
    
    def __init__(self):
        """Initialize the resume processor"""
        self.embeddings = self._initialize_embeddings()
        self.llm = self._initialize_llm()
        self.memory = ConversationBufferMemory(memory_key="chat_history", return_messages=True)
        self.resume_text = None
        self.structured_data = None
        self.vectorstore = None
        self.retriever = None
        self.qa_chain = None
    
    def _initialize_embeddings(self):
        """Initialize semantic embeddings model"""
        try:
            return HuggingFaceEmbeddings(
                model_name="sentence-transformers/all-MiniLM-L6-v2",
                model_kwargs={'device': 'cpu'}
            )
        except Exception as e:
            print(f"Error initializing embeddings: {e}")
            return None
    
    def _initialize_llm(self):
        """Initialize large language model for RAG"""
        try:
            return ChatGroq(
                model_name="llama-3.3-70b-versatile",  # Updated model name
                temperature=0.2,
                max_tokens=512
            )
        except Exception as e:
            print(f"Error initializing Groq LLM: {e}")
            print("Falling back to HuggingFace model...")
            try:
                from langchain.llms import HuggingFaceEndpoint
                return HuggingFaceEndpoint(
                    repo_id="google/flan-t5-large",
                    temperature=0.1,
                    max_length=512
                )
            except Exception as e2:
                print(f"Error initializing fallback LLM: {e2}")
                return None
    
    def load_resume_from_pdf(self, file_path):
        """Load and process a resume from a PDF file path"""
        try:
            with open(file_path, 'rb') as f:
                file_content = f.read()
            return self.process_resume_content(file_content)
        except Exception as e:
            print(f"Error loading PDF from path: {e}")
            return False
    
    def load_resume_from_bytes(self, file_content):
        """Load and process a resume from bytes content"""
        return self.process_resume_content(file_content)
    
    def process_resume_content(self, file_content):
        """Process resume content and build vector store"""
        try:
            # Extract text from PDF
            with pdfplumber.open(io.BytesIO(file_content)) as pdf:
                self.resume_text = "\n".join([page.extract_text() or "" for page in pdf.pages])
            
            if not self.resume_text.strip():
                print("No text content found in the PDF.")
                return False
            
            # Extract structured data
            self.structured_data = self._extract_structured_data(self.resume_text)
            
            # Create text chunks for vector db
            text_splitter = RecursiveCharacterTextSplitter(
                chunk_size=500,  # Smaller chunks for more precise retrieval
                chunk_overlap=100,
                length_function=len
            )
            chunks = text_splitter.split_text(self.resume_text)
            
            # Create metadata-enhanced documents
            documents = []
            for i, chunk in enumerate(chunks):
                # Determine section type for better retrieval context
                section_type = self._determine_section_type(chunk)
                documents.append(
                    Document(
                        page_content=chunk,
                        metadata={"chunk_id": i, "section_type": section_type}
                    )
                )
            
            # Create vector store if embeddings are available
            if self.embeddings:
                self.vectorstore = FAISS.from_documents(documents, self.embeddings)
                self.retriever = self.vectorstore.as_retriever(
                    search_kwargs={
                        "k": 4,  # Retrieve top 4 chunks
                        "fetch_k": 8  # Consider top 8 before returning 4 (diversity)
                    }
                )
                
                # Create QA chain with the retriever
                if self.llm:
                    self._setup_qa_chain()
                    return True
                else:
                    print("Warning: LLM not initialized. QA functionality will be limited.")
                    return False
            else:
                print("Warning: Embeddings model not initialized. Vector search will be unavailable.")
                return False
                
        except Exception as e:
            print(f"Error processing resume content: {e}")
            return False
    
    def _determine_section_type(self, text):
        """Determine the section type of a chunk of text"""
        text_lower = text.lower()
        if any(word in text_lower for word in ["education", "university", "college", "degree", "bachelor", "master", "phd"]):
            return "education"
        elif any(word in text_lower for word in ["experience", "work", "employment", "job", "position", "career"]):
            return "experience"
        elif any(word in text_lower for word in ["skill", "proficiency", "competency", "ability", "technology", "programming"]):
            return "skills"
        elif any(word in text_lower for word in ["contact", "email", "phone", "address", "linkedin"]):
            return "contact"
        elif any(word in text_lower for word in ["summary", "objective", "profile", "about"]):
            return "summary"
        else:
            return "other"
    
    
        
    def _setup_qa_chain(self):
        """Set up the QA chain with retriever and memory"""
        # Custom prompt that considers section types from metadata
        custom_prompt_template = """
        You are an AI assistant helping with resume analysis. Use the following pieces of resume content to answer the question.
        
        Resume sections:
        {context}
        
        Chat history:
        {chat_history}
        
        Question: {question}
        
        Important guidelines:
        - Be precise and factual about the information in the resume
        - If information is not found in the resume, say so clearly
        - Keep answers relevant to the resume content
        - Provide concrete examples from the resume when possible
        - Pay attention to the section types (education, experience, skills, etc.)
        
        Answer:
        """
        
        custom_prompt = PromptTemplate(
            input_variables=["context", "chat_history", "question"],
            template=custom_prompt_template
        )
        
        self.qa_chain = ConversationalRetrievalChain.from_llm(
            llm=self.llm,
            retriever=self.retriever,
            memory=self.memory,
            combine_docs_chain_kwargs={"prompt": custom_prompt},
            return_source_documents=True,
            verbose=True,
            output_key='answer'
        )
        # We might need to explicitly tell the memory which input/output keys to use
        self.qa_chain.memory.input_key = "question"
        self.qa_chain.memory.output_key = "answer"
    
    def _extract_structured_data(self, text):
        """Extract structured data from resume text using regex and heuristics"""
        resume_data = ResumeData()
        
        # Extract contact info
        # Name extraction - typically at the beginning of resume
        name_match = re.search(r'^([A-Z][a-z]+(?:\s[A-Z][a-z]+){1,3})', text, re.MULTILINE)
        if name_match:
            resume_data.contact_info.name = name_match.group(1)
        
        # Email extraction
        email_match = re.search(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)
        if email_match:
            resume_data.contact_info.email = email_match.group(0)
        
        # Phone extraction - various formats
        phone_patterns = [
            r'\(\d{3}\)\s*\d{3}[-.]?\d{4}',  # (123) 456-7890
            r'\d{3}[-.]?\d{3}[-.]?\d{4}',     # 123-456-7890 or 1234567890
            r'\+\d{1,2}\s*\(\d{3}\)\s*\d{3}[-.]?\d{4}'  # +1 (123) 456-7890
        ]
        for pattern in phone_patterns:
            phone_match = re.search(pattern, text)
            if phone_match:
                resume_data.contact_info.phone = phone_match.group(0)
                break
        
        # LinkedIn extraction
        linkedin_match = re.search(r'linkedin\.com/in/[a-zA-Z0-9_-]+', text)
        if linkedin_match:
            resume_data.contact_info.linkedin = "https://" + linkedin_match.group(0)
        
        # Extract education
        edu_patterns = [
            r'(?:Bachelor|BS|BA|B\.S\.|B\.A\.|Master|MS|MA|M\.S\.|M\.A\.|MBA|PhD|Ph\.D\.|Doctor)[^.]*?(?:University|College|Institute|School)[^.]*?\.', 
            r'(?:University|College|Institute|School)[^.]*?(?:Bachelor|BS|BA|B\.S\.|B\.A\.|Master|MS|MA|M\.S\.|M\.A\.|MBA|PhD|Ph\.D\.|Doctor)[^.]*?\.'
        ]
        
        for pattern in edu_patterns:
            for match in re.finditer(pattern, text, re.IGNORECASE):
                edu_text = match.group(0)
                
                # Find institution
                inst_match = re.search(r'(?:University|College|Institute|School)[^.,]*', edu_text, re.IGNORECASE)
                institution = inst_match.group(0) if inst_match else "Unknown Institution"
                
                # Find degree
                degree_pattern = r'(?:Bachelor|BS|BA|B\.S\.|B\.A\.|Master|MS|MA|M\.S\.|M\.A\.|MBA|PhD|Ph\.D\.|Doctor)[^.,]*'
                degree_match = re.search(degree_pattern, edu_text, re.IGNORECASE)
                degree = degree_match.group(0) if degree_match else None
                
                # Find dates
                date_match = re.search(r'\b(?:19|20)\d{2}\b(?:\s*-\s*(?:(?:19|20)\d{2}|Present|Current))?', edu_text)
                dates = date_match.group(0) if date_match else None
                
                resume_data.education.append(Education(
                    institution=institution.strip(),
                    degree=degree.strip() if degree else None,
                    dates=dates
                ))
        
        # Extract technical skills by looking for common skill terms
        skill_keywords = {
            "technical": [
                'Python', 'Java', 'JavaScript', 'SQL', 'C++', 'HTML', 'CSS', 'React', 
                'Angular', 'Node.js', 'AWS', 'Docker', 'Kubernetes', 'Git', 'Machine Learning',
                'Data Science', 'TensorFlow', 'PyTorch', 'Excel', 'PowerPoint', 'Word',
                'Agile', 'Scrum', 'Cloud', 'DevOps', 'API', 'REST', 'JSON', 'XML',
                'Linux', 'Unix', 'Windows', 'MacOS'
            ],
            "soft": [
                'Leadership', 'Communication', 'Teamwork', 'Problem Solving', 'Critical Thinking',
                'Time Management', 'Collaboration', 'Decision Making', 'Adaptability', 'Creativity',
                'Project Management', 'Negotiation', 'Conflict Resolution', 'Presentation'
            ]
        }
        
        # Extract skill sections
        skill_section_match = re.search(r'(?:SKILLS|Skills|skills|EXPERTISE|Expertise|COMPETENCIES|Competencies)[^\n]*(?:\n|.)+?(?=\n\s*\n|\Z)', text)
        if skill_section_match:
            skill_text = skill_section_match.group(0)
            
            # Check for each technical skill keyword
            for skill in skill_keywords["technical"]:
                if re.search(r'\b' + re.escape(skill) + r'\b', skill_text, re.IGNORECASE):
                    resume_data.skills.technical.append(skill)
            
            # Check for each soft skill keyword
            for skill in skill_keywords["soft"]:
                if re.search(r'\b' + re.escape(skill) + r'\b', skill_text, re.IGNORECASE):
                    resume_data.skills.soft.append(skill)
        
        # Extract work experience
        exp_section_match = re.search(r'(?:EXPERIENCE|Experience|WORK|Work|EMPLOYMENT|Employment)[^\n]*(?:\n|.)+?(?=(?:EDUCATION|Education|SKILLS|Skills|CERTIFICATIONS|Certifications)|\Z)', text)
        if exp_section_match:
            exp_text = exp_section_match.group(0)
            
            # Look for company/job patterns
            # Typically company and role are together with dates
            company_role_pattern = r'([A-Z][A-Za-z0-9\s&.,]+?)(?:,|\s+-\s+|\s+at\s+|\s+in\s+)([A-Za-z\s]+?)(?:\((\d{4}(?:\s*-\s*(?:\d{4}|Present|Current))?)\)|(\d{4}(?:\s*-\s*(?:\d{4}|Present|Current))?))'
            
            for match in re.finditer(company_role_pattern, exp_text):
                company = match.group(1).strip()
                title = match.group(2).strip()
                dates = match.group(3) if match.group(3) else match.group(4)
                
                # Extract responsibilities - text following the role until next role or end
                resp_text = exp_text[match.end():]
                end_pos = resp_text.find(company) if resp_text.find(company) > 0 else len(resp_text)
                resp_text = resp_text[:end_pos].strip()
                
                # Split responsibilities by bullet points or new lines
                responsibilities = []
                for line in re.split(r'•|\*|\n', resp_text):
                    cleaned_line = line.strip()
                    if cleaned_line and len(cleaned_line) > 10:  # Avoid very short lines
                        responsibilities.append(cleaned_line)
                
                resume_data.experience.append(Experience(
                    company=company,
                    title=title,
                    dates=dates,
                    responsibilities=responsibilities[:5]  # Limit to top 5 responsibilities
                ))
        
        return resume_data
    
    def ask_question(self, question):
        """Ask a question about the resume"""
        if not self.qa_chain:
            return "QA system not initialized. Please process a resume first."
        
        try:
            result = self.qa_chain({"question": question})
            return {
                "answer": result["answer"],
                "source_chunks": [doc.page_content for doc in result.get("source_documents", [])]
            }
        except Exception as e:
            return f"Error processing question: {str(e)}"
    
    def display_structured_data(self):
        """Return structured data in formatted text"""
        if not self.structured_data:
            return "No structured data available. Please process a resume first."
        
        output = []
        output.append("=== Resume Structured Data ===\n")
        
        # Contact info
        if self.structured_data.contact_info:
            output.append("CONTACT INFORMATION:")
            if self.structured_data.contact_info.name:
                output.append(f"Name: {self.structured_data.contact_info.name}")
            if self.structured_data.contact_info.email:
                output.append(f"Email: {self.structured_data.contact_info.email}")
            if self.structured_data.contact_info.phone:
                output.append(f"Phone: {self.structured_data.contact_info.phone}")
            if self.structured_data.contact_info.linkedin:
                output.append(f"LinkedIn: {self.structured_data.contact_info.linkedin}")
            output.append("")
        
        # Education
        if self.structured_data.education:
            output.append("EDUCATION:")
            for edu in self.structured_data.education:
                edu_str = f"• {edu.institution}"
                if edu.degree:
                    edu_str += f", {edu.degree}"
                if edu.dates:
                    edu_str += f" ({edu.dates})"
                output.append(edu_str)
            output.append("")
        
        # Skills
        if self.structured_data.skills:
            output.append("SKILLS:")
            if self.structured_data.skills.technical:
                output.append(f"Technical: {', '.join(self.structured_data.skills.technical)}")
            if self.structured_data.skills.soft:
                output.append(f"Soft: {', '.join(self.structured_data.skills.soft)}")
            output.append("")
        
        # Experience
        if self.structured_data.experience:
            output.append("EXPERIENCE:")
            for i, exp in enumerate(self.structured_data.experience):
                output.append(f"• {exp.title} at {exp.company} {f'({exp.dates})' if exp.dates else ''}")
                if exp.responsibilities:
                    for j, resp in enumerate(exp.responsibilities[:3]):  # Show top 3 responsibilities
                        output.append(f"  - {resp}")
                if i < len(self.structured_data.experience) - 1:
                    output.append("")
        
        return "\n".join(output)

# Example usage
def example_usage():
    print("Initializing Resume Processor...")
    processor = ResumeProcessor()
    
    # Example with direct file path
    print("\nProcessing resume...")
    success = processor.load_resume_from_pdf("path/to/your/resume.pdf")
    
    if success:
        print("\nStructured Data Extracted:")
        print(processor.display_structured_data())
        
        print("\nAsking questions about the resume:")
        questions = [
            "What skills does this candidate have?",
            "Where did this person go to school?",
            "What was their most recent job?",
            "What are their key achievements?"
        ]
        
        for question in questions:
            print(f"\nQ: {question}")
            result = processor.ask_question(question)
            if isinstance(result, dict):
                print(f"A: {result['answer']}")
                print("\nSource chunks:")
                for i, chunk in enumerate(result['source_chunks'][:2]):  # Show top 2 source chunks
                    print(f"Chunk {i+1}: {chunk[:100]}...")
            else:
                print(f"A: {result}")
    else:
        print("Failed to process resume.")

# Uncomment to run the example
# example_usage()

# Interactive demo
def interactive_demo():
    processor = ResumeProcessor()
    
    print("Resume Chatbot (Vector DB + Groq RAG) Demo")
    print("------------------------------------------")
    print("1. First, enter the path to your PDF resume file")
    print("2. Then ask questions about the resume")
    print("Type 'show data' to see extracted structured data")
    print("Type 'exit' to quit")
    
    pdf_path = input("\nEnter resume PDF path: ")
    
    if processor.load_resume_from_pdf(pdf_path):
        print("Resume processed successfully!")
        print("You can now ask questions about the resume.")
        
        while True:
            query = input("\nAsk a question (or 'show data'/'exit'): ")
            
            if query.lower() == 'exit':
                print("Exiting...")
                break
            elif query.lower() == 'show data':
                print("\n" + processor.display_structured_data() + "\n")
            else:
                result = processor.ask_question(query)
                if isinstance(result, dict):
                    print(f"Answer: {result['answer']}")
                else:
                    print(f"Answer: {result}")
    else:
        print(f"Failed to process resume at: {pdf_path}")

interactive_demo()
print("hi")
````
