```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
print("Libraries imported successfully.")
```

```python
# Load environment variables from .env file
if load_dotenv():
    print("Environment variables loaded from .env file.")
else:
    print("No .env file found or it's empty. Make sure GROQ_API_KEY is set in a .env file in the same directory as this notebook.")

groq_api_key = os.getenv("GROQ_API_KEY")

chat_model = None

if not groq_api_key:
    print("ERROR: GROQ_API_KEY not found in environment variables.")
    print("Please ensure it's set in your .env file (e.g., GROQ_API_KEY='your_key_here').")
else:
    print("GROQ_API_KEY loaded successfully.")
    # Initialize the Groq Chat client
    try:
        chat_model = ChatGroq(
            groq_api_key=groq_api_key,
            model_name="llama3-8b-8192", 
            temperature=0.7 # Adjust for creativity (higher) vs. factuality (lower)
        )
        print(f"Groq Chat client initialized with model: {chat_model.model_name}")
    except Exception as e:
        print(f"Error initializing Groq client: {e}")
        print("Please check your API key and model name.")
```

```python
qa_chain = None

if chat_model: # Proceed only if chat_model was initialized successfully
    # Define a prompt template
    prompt_template = ChatPromptTemplate.from_messages([
        ("system", "You are a helpful assistant. Answer the user's questions clearly and concisely."),
        ("human", "{question}") 
    ])

    # Define the output parser
    output_parser = StrOutputParser()

    # Create the chain: Prompt -> LLM -> Output Parser
    qa_chain = prompt_template | chat_model | output_parser

    print("LangChain QA chain created successfully.")
else:
    print("Skipping LangChain pipeline creation as Groq client (chat_model) was not initialized.")
```

```python
if qa_chain:
    sample_question = "What is the colour of the sky?"
    print(f"\nTesting with sample question: '{sample_question}'")
    print("Bot: Thinking...")
    try:
        response = qa_chain.invoke({"question": sample_question})
        print("\nLLM Response:")
        print(response)
    except Exception as e:
        print(f"Error invoking QA chain: {e}")
else:
    print("Cannot test QA chain as it was not created (chat_model or chain setup failed).")
```

# Part 2: Advanced Resume RAG Chatbot

```python
import io
import pdfplumber
import re
from typing import List, Dict, Any, Optional
from pydantic import BaseModel, Field
import os

try:
    from langchain.embeddings import HuggingFaceEmbeddings
    from langchain.vectorstores import FAISS
except ImportError:
    print("Installing required packages...")
    !pip install langchain-groq faiss-cpu sentence-transformers
    import langchain_groq
    from langchain.embeddings import HuggingFaceEmbeddings
    from langchain.vectorstores import FAISS

from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationalRetrievalChain
from langchain.prompts import PromptTemplate
from langchain_groq import ChatGroq
from langchain.schema import Document
```

```python
class ContactInfo(BaseModel):
    """Contact information from a resume"""
    name: Optional[str] = Field(None, description="Full name of the candidate")
    email: Optional[str] = Field(None, description="Email address of the candidate")
    phone: Optional[str] = Field(None, description="Phone number of the candidate")
    location: Optional[str] = Field(None, description="Location/address of the candidate")
    linkedin: Optional[str] = Field(None, description="LinkedIn profile URL")
    
class Education(BaseModel):
    """Education information from a resume"""
    institution: str = Field(..., description="Name of educational institution")
    degree: Optional[str] = Field(None, description="Degree obtained or pursuing")
    field_of_study: Optional[str] = Field(None, description="Field or major of study")
    dates: Optional[str] = Field(None, description="Start and end dates of education")
    gpa: Optional[str] = Field(None, description="GPA if mentioned")

class Experience(BaseModel):
    """Work experience information from a resume"""
    company: str = Field(..., description="Company or organization name")
    title: str = Field(..., description="Job title or position")
    dates: Optional[str] = Field(None, description="Employment duration")
    responsibilities: List[str] = Field(default_factory=list, description="Key responsibilities and achievements")

class Skills(BaseModel):
    """Skills information from a resume"""
    technical: List[str] = Field(default_factory=list, description="Technical skills (programming, tools, etc.)")
    soft: List[str] = Field(default_factory=list, description="Soft skills (communication, leadership, etc.)")
    languages: List[str] = Field(default_factory=list, description="Language proficiencies")

class ResumeData(BaseModel):
    """Comprehensive resume data structure"""
    contact_info: ContactInfo = Field(default_factory=ContactInfo)
    summary: Optional[str] = Field(None, description="Professional summary or objective statement")
    education: List[Education] = Field(default_factory=list)
    experience: List[Experience] = Field(default_factory=list)
    skills: Skills = Field(default_factory=Skills)
    certifications: List[str] = Field(default_factory=list, description="Professional certifications")
```

```python
class ResumeProcessor:
    """Core class for processing resumes with advanced retrieval capabilities"""
    
    def __init__(self):
        """Initialize the resume processor"""
        self.embeddings = self._initialize_embeddings()
        self.llm = chat_model
        self.memory = ConversationBufferMemory(memory_key="chat_history", return_messages=True, output_key='answer')
        self.resume_text = None
        self.structured_data = None
        self.vectorstore = None
        self.retriever = None
        self.qa_chain = None
    
    def _initialize_embeddings(self):
        """Initialize semantic embeddings model"""
        try:
            return HuggingFaceEmbeddings(
                model_name="sentence-transformers/all-MiniLM-L6-v2",
                model_kwargs={'device': 'cpu'}
            )
        except Exception as e:
            print(f"Error initializing embeddings: {e}")
            return None
    
    def load_resume_from_pdf(self, file_path):
        """Load and process a resume from a PDF file path"""
        try:
            with open(file_path, 'rb') as f:
                file_content = f.read()
            return self.process_resume_content(file_content)
        except Exception as e:
            print(f"Error loading PDF from path: {e}")
            return False
    
    def load_resume_from_bytes(self, file_content):
        """Load and process a resume from bytes content"""
        return self.process_resume_content(file_content)
    
    def process_resume_content(self, file_content):
        """Process resume content and build vector store"""
        try:
            with pdfplumber.open(io.BytesIO(file_content)) as pdf:
                self.resume_text = "\n".join([page.extract_text() or "" for page in pdf.pages])
            
            if not self.resume_text.strip():
                print("No text content found in the PDF.")
                return False
            
            self.structured_data = self._extract_structured_data(self.resume_text)
            
            text_splitter = RecursiveCharacterTextSplitter(
                chunk_size=500,  # Smaller chunks for more precise retrieval
                chunk_overlap=100,
                length_function=len
            )
            chunks = text_splitter.split_text(self.resume_text)
            
            documents = []
            for i, chunk in enumerate(chunks):
                documents.append(Document(page_content=chunk))
            
            self.vectorstore = FAISS.from_documents(documents, self.embeddings)
            self.retriever = self.vectorstore.as_retriever(search_kwargs={"k": 4})
            self._setup_qa_chain()
            return True
        
        except Exception as e:
            print(f"Error processing resume: {e}")
            return False

    def _setup_qa_chain(self):
        """Set up the QA chain with retriever and memory"""
        # Custom prompt that considers section types from metadata
        custom_prompt_template = """
        You are an AI assistant helping with resume analysis. Use the following pieces of resume content to answer the question.
        
        Resume sections:
        {context}
        
        Chat history:
        {chat_history}
        
        Question: {question}
        
        Important guidelines:
        - Be precise and factual about the information in the resume
        - If information is not found in the resume, say so clearly
        - Keep answers relevant to the resume content
        - Provide concrete examples from the resume when possible
        - Pay attention to the section types (education, experience, skills, etc.)
        
        Answer:
        """
        
        custom_prompt = PromptTemplate(
            input_variables=["context", "chat_history", "question"],
            template=custom_prompt_template
        )
        
        self.qa_chain = ConversationalRetrievalChain.from_llm(
            llm=self.llm,
            retriever=self.retriever,
            memory=self.memory,
            combine_docs_chain_kwargs={"prompt": custom_prompt},
            return_source_documents=True,
            verbose=True,
            output_key='answer'
        )
        # We might need to explicitly tell the memory which input/output keys to use
        self.qa_chain.memory.input_key = "question"
        self.qa_chain.memory.output_key = "answer"
    
    def _extract_structured_data(self, text):
        """Extract structured data from resume text using regex and heuristics"""
        resume_data = ResumeData()
        
        # Extract contact info
        # Name extraction - typically at the beginning of resume
        name_match = re.search(r'^([A-Z][a-z]+(?:\s[A-Z][a-z]+){1,3})', text, re.MULTILINE)
        if name_match:
            resume_data.contact_info.name = name_match.group(1)
        
        # Email extraction
        email_match = re.search(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', text)
        if email_match:
            resume_data.contact_info.email = email_match.group(0)
        
        # Phone extraction - various formats
        phone_patterns = [
            r'\(\d{3}\)\s*\d{3}[-.]?\d{4}',  # (123) 456-7890
            r'\d{3}[-.]?\d{3}[-.]?\d{4}',     # 123-456-7890 or 1234567890
            r'\+\d{1,2}\s*\(\d{3}\)\s*\d{3}[-.]?\d{4}'  # +1 (123) 456-7890
        ]
        for pattern in phone_patterns:
            phone_match = re.search(pattern, text)
            if phone_match:
                resume_data.contact_info.phone = phone_match.group(0)
                break
        
        # LinkedIn extraction
        linkedin_match = re.search(r'linkedin\.com/in/[a-zA-Z0-9_-]+', text)
        if linkedin_match:
            resume_data.contact_info.linkedin = "https://" + linkedin_match.group(0)
        
        # Extract education
        edu_patterns = [
            r'(?:Bachelor|BS|BA|B\.S\.|B\.A\.|Master|MS|MA|M\.S\.|M\.A\.|MBA|PhD|Ph\.D\.|Doctor)[^.]*?(?:University|College|Institute|School)[^.]*?\.', 
            r'(?:University|College|Institute|School)[^.]*?(?:Bachelor|BS|BA|B\.S\.|B\.A\.|Master|MS|MA|M\.S\.|M\.A\.|MBA|PhD|Ph\.D\.|Doctor)[^.]*?\.'
        ]
        
        for pattern in edu_patterns:
            for match in re.finditer(pattern, text, re.IGNORECASE):
                edu_text = match.group(0)
                
                # Find institution
                inst_match = re.search(r'(?:University|College|Institute|School)[^.,]*', edu_text, re.IGNORECASE)
                institution = inst_match.group(0) if inst_match else "Unknown Institution"
                
                # Find degree
                degree_pattern = r'(?:Bachelor|BS|BA|B\.S\.|B\.A\.|Master|MS|MA|M\.S\.|M\.A\.|MBA|PhD|Ph\.D\.|Doctor)[^.,]*'
                degree_match = re.search(degree_pattern, edu_text, re.IGNORECASE)
                degree = degree_match.group(0) if degree_match else None
                
                # Find dates
                date_match = re.search(r'\b(?:19|20)\d{2}\b(?:\s*-\s*(?:(?:19|20)\d{2}|Present|Current))?', edu_text)
                dates = date_match.group(0) if date_match else None
                
                resume_data.education.append(Education(
                    institution=institution.strip(),
                    degree=degree.strip() if degree else None,
                    dates=dates
                ))
        
        # Extract technical skills by looking for common skill terms
        skill_keywords = {
            "technical": [
                'Python', 'Java', 'JavaScript', 'SQL', 'C++', 'HTML', 'CSS', 'React', 
                'Angular', 'Node.js', 'AWS', 'Docker', 'Kubernetes', 'Git', 'Machine Learning',
                'Data Science', 'TensorFlow', 'PyTorch', 'Excel', 'PowerPoint', 'Word',
                'Agile', 'Scrum', 'Cloud', 'DevOps', 'API', 'REST', 'JSON', 'XML',
                'Linux', 'Unix', 'Windows', 'MacOS'
            ],
            "soft": [
                'Leadership', 'Communication', 'Teamwork', 'Problem Solving', 'Critical Thinking',
                'Time Management', 'Collaboration', 'Decision Making', 'Adaptability', 'Creativity',
                'Project Management', 'Negotiation', 'Conflict Resolution', 'Presentation'
            ]
        }
        
        # Extract skill sections
        skill_section_match = re.search(r'(?:SKILLS|Skills|skills|EXPERTISE|Expertise|COMPETENCIES|Competencies)[^\n]*(?:\n|.)+?(?=\n\s*\n|\Z)', text)
        if skill_section_match:
            skill_text = skill_section_match.group(0)
            
            # Check for each technical skill keyword
            for skill in skill_keywords["technical"]:
                if re.search(r'\b' + re.escape(skill) + r'\b', skill_text, re.IGNORECASE):
                    resume_data.skills.technical.append(skill)
            
            # Check for each soft skill keyword
            for skill in skill_keywords["soft"]:
                if re.search(r'\b' + re.escape(skill) + r'\b', skill_text, re.IGNORECASE):
                    resume_data.skills.soft.append(skill)
        
        # Extract work experience
        exp_section_match = re.search(r'(?:EXPERIENCE|Experience|WORK|Work|EMPLOYMENT|Employment)[^\n]*(?:\n|.)+?(?=(?:EDUCATION|Education|SKILLS|Skills|CERTIFICATIONS|Certifications)|\Z)', text)
        if exp_section_match:
            exp_text = exp_section_match.group(0)
            
            # Look for company/job patterns
            company_role_pattern = r'([A-Z][A-Za-z0-9\s&.,]+?)(?:,|\s+-\s+|\s+at\s+|\s+in\s+)([A-Za-z\s]+?)(?:\((\d{4}(?:\s*-\s*(?:\d{4}|Present|Current))?)\)|(\d{4}(?:\s*-\s*(?:\d{4}|Present|Current))?))'
            
            for match in re.finditer(company_role_pattern, exp_text):
                company = match.group(1).strip()
                title = match.group(2).strip()
                dates = match.group(3) if match.group(3) else match.group(4)
                
                # Extract responsibilities - text following the role until next role or end
                resp_text = exp_text[match.end():]
                end_pos = resp_text.find(company) if resp_text.find(company) > 0 else len(resp_text)
                resp_text = resp_text[:end_pos].strip()
                
                # Split responsibilities by bullet points or new lines
                responsibilities = []
                for line in re.split(r'•|\*|\n', resp_text):
                    cleaned_line = line.strip()
                    if cleaned_line and len(cleaned_line) > 10:  # Avoid very short lines
                        responsibilities.append(cleaned_line)
                
                resume_data.experience.append(Experience(
                    company=company,
                    title=title,
                    dates=dates,
                    responsibilities=responsibilities[:5]  # Limit to top 5 responsibilities
                ))
        
        return resume_data

    def ask_question(self, question):
        """Ask a question about the resume"""
        if not self.qa_chain:
            return "QA system not initialized. Please process a resume first."
        
        try:
            result = self.qa_chain({"question": question})
            return {
                "answer": result["answer"],
                "source_chunks": [doc.page_content for doc in result.get("source_documents", [])]
            }
        except Exception as e:
            return f"Error processing question: {str(e)}"
    
    def display_structured_data(self):
        """Return structured data in formatted text"""
        if not self.structured_data:
            return "No structured data available. Please process a resume first."
        
        output = []
        output.append("=== Resume Structured Data ===\n")
        
        # Contact info
        if self.structured_data.contact_info:
            output.append("CONTACT INFORMATION:")
            if self.structured_data.contact_info.name:
                output.append(f"Name: {self.structured_data.contact_info.name}")
            if self.structured_data.contact_info.email:
                output.append(f"Email: {self.structured_data.contact_info.email}")
            if self.structured_data.contact_info.phone:
                output.append(f"Phone: {self.structured_data.contact_info.phone}")
            if self.structured_data.contact_info.linkedin:
                output.append(f"LinkedIn: {self.structured_data.contact_info.linkedin}")
            output.append("")
        
        # Education
        if self.structured_data.education:
            output.append("EDUCATION:")
            for edu in self.structured_data.education:
                edu_str = f"• {edu.institution}"
                if edu.degree:
                    edu_str += f", {edu.degree}"
                if edu.dates:
                    edu_str += f" ({edu.dates})"
                output.append(edu_str)
            output.append("")
        
        # Skills
        if self.structured_data.skills:
            output.append("SKILLS:")
            if self.structured_data.skills.technical:
                output.append(f"Technical: {', '.join(self.structured_data.skills.technical)}")
            if self.structured_data.skills.soft:
                output.append(f"Soft: {', '.join(self.structured_data.skills.soft)}")
            output.append("")
        
        # Experience
        if self.structured_data.experience:
            output.append("EXPERIENCE:")
            for i, exp in enumerate(self.structured_data.experience):
                output.append(f"• {exp.title} at {exp.company} {f'({exp.dates})' if exp.dates else ''}")
                if exp.responsibilities:
                    for j, resp in enumerate(exp.responsibilities[:3]):  # Show top 3 responsibilities
                        output.append(f"  - {resp}")
                if i < len(self.structured_data.experience) - 1:
                    output.append("")
        
        return "\n".join(output)

print("ResumeProcessor class defined.")
```

```python
# Example usage
def example_usage():
    print("Initializing Resume Processor...")
    processor = ResumeProcessor()
    
    # Example with direct file path
    print("\nProcessing resume...")
    success = processor.load_resume_from_pdf("path/to/your/resume.pdf")
    
    if success:
        print("\nStructured Data Extracted:")
        print(processor.display_structured_data())
        
        print("\nAsking questions about the resume:")
        questions = [
            "What skills does this candidate have?",
            "Where did this person go to school?",
            "What was their most recent job?",
            "What are their key achievements?"
        ]
        
        for question in questions:
            print(f"\nQ: {question}")
            result = processor.ask_question(question)
            if isinstance(result, dict):
                print(f"A: {result['answer']}")
                # print("\nSource chunks:") # Uncomment for debugging
                # for i, chunk in enumerate(result['source_chunks'][:2]):  # Show top 2 source chunks
                #     print(f"Chunk {i+1}: {chunk[:100]}...")
            else:
                print(f"A: {result}")
    else:
        print("Failed to process resume.")

# Uncomment to run the example
# example_usage()
```

```python
# Interactive demo
def interactive_demo():
    processor = ResumeProcessor()
    
    print("Resume Chatbot (Vector DB + Groq RAG) Demo")
    print("1. First, enter the path to your PDF resume file")
    print("2. Then ask questions about the resume")
    print("Type 'show data' to see extracted structured data")
    print("Type 'exit' to quit")
    
    pdf_path = input("\nEnter resume PDF path: ").strip().strip("'\"")
    
    if processor.load_resume_from_pdf(pdf_path):
        print("Resume processed successfully!")
        print("You can now ask questions about the resume.")
        
        while True:
            query = input("\nAsk a question (or 'show data'/'exit'): ")
            
            if query.lower() == 'exit':
                print("Exiting...")
                break
            elif query.lower() == 'show data':
                print("\n" + processor.display_structured_data() + "\n")
            else:
                result = processor.ask_question(query)
                if isinstance(result, dict):
                    print(f"Answer: {result['answer']}")
                else:
                    print(f"Answer: {result}")
    else:
        print(f"Failed to process resume at: {pdf_path}")
```
